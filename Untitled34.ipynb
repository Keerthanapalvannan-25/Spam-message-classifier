```python
import pandas as pd

df = pd.read_csv('spam.csv', encoding='latin-1')

df = df.drop(columns=['Unnamed: 2', 'Unnamed: 3', 'Unnamed: 4'], errors='ignore')

df.columns = ['label', 'message']

df['label'] = df['label'].map({'ham': 0, 'spam': 1})

print("Step 1 Completed: Data Loaded and Cleaned!")
print(f"Total messages: {len(df)}")
print(df.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

X_train, X_test, y_train, y_test = train_test_split(
    df['message'],
    df['label'],
    test_size=0.2,
    random_state=42
)

tfidf = TfidfVectorizer(stop_words='english', lowercase=True)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Step 2 Completed!")
print(f"Training set size: {X_train_tfidf.shape[0]} messages")
print(f"Number of unique words (features): {X_train_tfidf.shape[1]}")
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

spam_model = MultinomialNB()
spam_model.fit(X_train_tfidf, y_train)

y_pred = spam_model.predict(X_test_tfidf)

accuracy = accuracy_score(y_test, y_pred)
print("Step 3 Completed!")
print(f"Model Accuracy: {accuracy:.2%}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=['Ham', 'Spam']))
```

```python
import pickle

def predict_spam(message):
    vectorized_msg = tfidf.transform([message])
    prediction = spam_model.predict(vectorized_msg)
    return "SPAM 🚨" if prediction[0] == 1 else "HAM (Normal) ✅"

print(f"Message 1: {predict_spam('Congratulations! You won a free gift card. Click here to claim.')}")
print(f"Message 2: {predict_spam('Hey, are we still meeting for lunch today?')}")

with open('spam_model.pkl', 'wb') as f:
    pickle.dump(spam_model, f)

with open('spam_tfidf.pkl', 'wb') as f:
    pickle.dump(tfidf, f)

print("\nStep 4 Completed: Model and Vectorizer saved!")
```
